# Prompt A/B 试验骨架

```python
from mysql_agent.prompts.ab_test import PromptABTest, ABVariant

ab = PromptABTest('rag_tone', [
    ABVariant('concise', '简要回答：{q}', weight=1),
    ABVariant('detailed', '详细分步骤回答：{q}', weight=1),
])

def execute(variant, payload):
    return variant.prompt_template.format(q=payload)

def score(out):
    return 1.0 if '回答' in out else 0.5

for q in ['死锁怎么查', '慢查询怎么优化', '复制延迟']: 
    ab.run_once(q, execute, score)
ab.summary()
```
